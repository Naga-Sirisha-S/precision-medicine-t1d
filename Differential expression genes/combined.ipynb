```python
import pandas as pd

# File paths (your uploaded files)
file1_path = "D:\BIOINFORMATICS\OMICS LOGIC\PROJECTS\combined_data\GSE123658-GPL18573_series_matrix.xlsx"
file2_path = "D:\BIOINFORMATICS\OMICS LOGIC\PROJECTS\combined_data\GSE123658-GPL20301_series_matrix.xlsx"

# Load both Excel files
df1 = pd.read_excel(file1_path, header=None)
df2 = pd.read_excel(file2_path, header=None)
```

```python
df1.head()
```

```python
# Rows to extract
rows_needed = ["!Sample_title", "!Sample_characteristics_ch1"]

# Extract required rows from file1
extract1 = df1[df1[0].isin(rows_needed)]

# Extract required rows from file2
extract2 = df2[df2[0].isin(rows_needed)]
```

```python
# Combine both side-by-side
combined = pd.concat([extract1, extract2], axis=1)
```

```python
combined.head()
```

```python
# Save into new Excel file
output_path = "D:/BIOINFORMATICS/OMICS LOGIC/PROJECTS/combined_data/combined_sample_info.xlsx"
combined.to_excel(output_path, index=False, header=False)

print("Combined Excel file created successfully!")
print("Saved at:", output_path)
```

```python
combined.shape
```

```python
combined -- spliting
```

```python
# Find the row where first column is "!Sample_title"
sample_idx = combined[combined.iloc[:, 0] == "!Sample_title"].index

# Extract that row
sample_row = combined.loc[sample_idx[0]]

# Create list to store separated rows
rows = []

# Loop through all values in that row (excluding first column)
for val in sample_row[1:]:
    if isinstance(val, str) and "ID:" in val:
        # Split into subject type and ID
        part1, part2 = val.split("ID:")

        rows.append({
            "Subject_Type": part1.strip() + "ID",
            "ID_Value": part2.strip()
        })
    else:
        rows.append({
            "Subject_Type": val,
            "ID_Value": ""
        })

# Convert into DataFrame
output_df = pd.DataFrame(rows)
```

```python
#Step 2: Extract sample names from read_count1 columns
# -------------------------------

sample_names = read_count1.columns.tolist()

# -------------------------------
# Step 3: Create mapping: Sample_ID → Group_info
# -------------------------------

sample_to_group = dict(
    zip(combined_data1["Sample_ID"], combined_data1["Group_info"])
)

# -------------------------------
# Step 4: Replace sample names with Group_info
# -------------------------------

new_columns = []

for sample in sample_names:
    if sample in sample_to_group:
        new_columns.append(sample_to_group[sample])
    else:
        new_columns.append(sample)  # keep unchanged if no match

read_count1.columns = new_columns
```
